## 2.1 Krachten in één richting (1D)

In eerdere lessen heb je vooral gewerkt met krachten die in één richting werken,  
bijvoorbeeld horizontaal of verticaal. Dit noemen we eendimensionale krachten (1D).

Kijk naar het voorbeeld in Figuur 1 (zie onder):  
Een auto rijdt weg van een stoplicht. De auto versnelt horizontaal,  
maar blijft verticaal stil staan — hij zweeft niet de lucht in  
en zakt ook niet door de grond.

Dat betekent dat er in verticale richting een krachtenevenwicht is:  
de normaalkracht van de grond heft precies het gewicht van de auto op.

We zeggen dan:
- In de **y-richting** (1e wet van Newton):  
  $ \sum F_y = 0 $ → geen beweging omhoog of omlaag.
- In de **x-richting** (2e wet van Newton):  
  $ \sum F_x = ma $ → want de auto versnelt.

Wanneer we vectoren bij elkaar optellen krijgen we:
- In de y-richting:  
  $ \sum F_y = F_n - F_z $
- In de x-richting:  
  $ \sum F_x = F_{\text{auto}} - F_w $


## 2.2 Krachten in meerdere richtingen (2D)

Zodra krachten niet meer netjes in dezelfde richting liggen,  
hebben we te maken met **tweedimensionale krachten (2D)**.

Stel je voor dat je van punt A naar B loopt (vector $\vec{AB}$),  
en dan van B naar C (vector $\vec{BC}$).  
De totale verplaatsing is dan van A naar C (vector $\vec{AC}$).  
Dit is het idee achter het optellen van vectoren via de **kop-staartmethode**.

In Figuur 2 zie je hoe dit werkt.  
De **parallellogrammethode** is eigenlijk de kop-staartmethode,  
maar dan twee keer toegepast.

We kunnen deze methode ook gebruiken om een kracht te **ontbinden** in twee richtingen:  
bijvoorbeeld een schuine kracht splitsen in een horizontale en verticale component.

Dit is handig omdat we dan weer met eenvoudige 1D-vergelijkingen kunnen rekenen.


```python
%pip install jupyterquiz



:::{quiz} Krachten in één richting – Auto bij stoplicht
:shuffle: false
:quizId: krachten-1d-auto

### Vraag 1
De auto rijdt weg van een stoplicht.

De 1e wet van Newton zegt dat een voorwerp waarop geen resulterende kracht ($F_\text{res} = 0$) werkt, in rust blijft of met constante snelheid beweegt.

De 2e wet zegt dat als er een kracht werkt, $F_\text{res} = ma$, het versnelt.

**Welke twee uitspraken kloppen?**

- [ ] A) In de horizontale richting geldt: $F_\text{res} = 0$
- [x] B) In de horizontale richting geldt: $F_\text{res} = ma$
- [x] C) In de verticale richting geldt: $F_\text{res} = 0$
- [ ] D) In de verticale richting geldt: $F_\text{res} = ma$

---

### Vraag 2
De krachten op de auto zijn: zwaartekracht ($F_z$), normaalkracht ($F_n$), aandrijfkracht ($F_\text{auto}$), en rolwrijving ($F_w$).  

**Welke uitspraak is juist over hun richtingen?**

- [ ] A) Horizontaal: $F_\text{auto}, F_n$ & Verticaal: $F_z, F_w$
- [x] B) Horizontaal: $F_\text{auto}, F_w$ & Verticaal: $F_z, F_n$
- [ ] C) Horizontaal: $F_\text{auto}$ & Verticaal: $F_z$

---

### Vraag 3
De resulterende kracht $F_\text{res}$ ontstaat uit de optelsom van alle krachten $\sum F$ in die richting.

In verticale richting is: $\sum F_y = 0$,  
In horizontale richting is: $\sum F_x = ma$

**Kies voor beide stellingen uit: groter dan, kleiner dan, gelijk aan.**

**De zwaartekracht is ... de normaalkracht**

- [x] gelijk aan
- [ ] groter dan
- [ ] kleiner dan

**De aandrijfkracht is ... de wrijvingskracht**

- [x] groter dan
- [ ] gelijk aan
- [ ] kleiner dan

---

### Vraag 4
Welke wetten van Newton gelden hier?

- [x] In y-richting: 1e wet → $\sum F_y = 0$
- [x] In x-richting: 2e wet → $\sum F_x = ma$
- [ ] In y-richting: 2e wet → $\sum F_y = ma$
- [ ] In x-richting: 1e wet → $\sum F_x = 0$
:::
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

question = widgets.Label("Welke 2 uitspraken kloppen?")

options = [
    widgets.Checkbox(value=False, description="A) In de horizontale richting geldt F_res = 0"),
    widgets.Checkbox(value=True,  description="B) In de horizontale richting geldt F_res = ma"),
    widgets.Checkbox(value=True,  description="C) In de verticale richting geldt F_res = 0"),
    widgets.Checkbox(value=False, description="D) In de verticale richting geldt F_res = ma")
]

button = widgets.Button(description="Controleer antwoord")
output = widgets.Output()

def check_answer(b):
    with output:
        clear_output()
        correct = [options[1].value, options[2].value] == [True, True]
        print("✅ Correct!" if correct else "❌ Niet helemaal goed, probeer opnieuw.")

button.on_click(check_answer)

display(question, *options, button, output)
```
